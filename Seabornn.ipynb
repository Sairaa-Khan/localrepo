```python
import seaborn as sns
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
flight_df=sns.load_dataset('flights')
flight_df.head()
```

```python
flight_df=flight_df.pivot(index="month", columns="year", values="passengers")

flight_df.head()
```

```python
plt.figure(figsize=(14,8))
sns.heatmap(flight_df,
            annot=True,
            fmt='d',
            linecolor='k',
            linewidth='3',
            # cbar=False

            cmap='Reds')
# print(ax)
```

```python
# Horizontal Cbar
grid_kws = {"height_ratios": (.5, .05) ,"hspace":.4}
f,(ax,cbar_ax)=plt.subplots(2,gridspec_kw=grid_kws)
ax=sns.heatmap(flight_df, cbar_kws={"orientation":"horizontal"},ax=ax,cbar_ax=cbar_ax)
```
